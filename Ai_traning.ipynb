````python

# 1. Libraries setup
!pip install flask flask-cors ollama -q

import os
from flask import Flask, request, jsonify
from flask_cors import CORS
import ollama
import threading
from google.colab import output

app = Flask(__name__)
CORS(app)

# 2. HTML, CSS & JS Code
html_code = """
<!DOCTYPE html>
<html lang="en">
<head>
    <meta charset="UTF-8">
    <meta name="viewport" content="width=device-width, initial-scale=1.0">
    <title>Qwen Code AI Assistant</title>
    <style>
        * { margin: 0; padding: 0; box-sizing: border-box; font-family: 'Segoe UI', Tahoma, Geneva, Verdana, sans-serif; }
        body { background-color: #0b0f19; color: #e2e8f0; display: flex; flex-direction: column; height: 100vh; }
        header { background-color: #111827; padding: 15px; text-align: center; border-bottom: 1px solid #1f2937; color: #10b981; font-size: 20px; font-weight: bold; }
        #chat-container { flex: 1; padding: 20px; overflow-y: auto; display: flex; flex-direction: column; gap: 15px; }
        .message { max-width: 80%; padding: 12px 16px; border-radius: 12px; line-height: 1.5; font-size: 15px; word-wrap: break-word; }
        .user { background-color: #047857; color: white; align-self: flex-end; border-bottom-right-radius: 2px; }
        .bot { background-color: #1f2937; color: #e2e8f0; align-self: flex-start; border-bottom-left-radius: 2px; border: 1px solid #374151; }
        pre { background-color: #030712; padding: 10px; border-radius: 6px; overflow-x: auto; margin-top: 5px; border: 1px solid #1f2937; color: #34d399; font-family: monospace; }
        #input-container { padding: 15px; background-color: #111827; display: flex; gap: 10px; border-top: 1px solid #1f2937; }
        input { flex: 1; padding: 12px; border-radius: 8px; border: 1px solid #374151; background-color: #1f2937; color: white; font-size: 15px; outline: none; }
        input:focus { border-color: #10b981; }
        button { background-color: #10b981; color: white; border: none; padding: 0 20px; border-radius: 8px; font-weight: bold; cursor: pointer; transition: 0.2s; }
        button:hover { background-color: #059669; }
    </style>
</head>
<body>

    <header>⚡ Qwen2.5-Coder Expert AI</header>
    <div id="chat-container"></div>
    <div id="input-container">
        <input type="text" id="user-input" placeholder="Apna code ya Chartink sawal likhein..." onkeypress="handleKeyPress(event)">
        <button onclick="sendMessage()">Send</button>
    </div>

    <script>
        const chatContainer = document.getElementById('chat-container');
        const userInput = document.getElementById('user-input');

        function appendMessage(text, sender) {
            const msgDiv = document.createElement('div');
            msgDiv.classList.add('message', sender);

            if (text.includes('```')) {
                let parts = text.split('```');
                for(let i=0; i<parts.length; i++) {
                    if(i % 2 === 1) {
                        let pre = document.createElement('pre');
                        pre.innerText = parts[i].trim();
                        msgDiv.appendChild(pre);
                    } else {
                        let span = document.createElement('span');
                        span.innerText = parts[i];
                        msgDiv.appendChild(span);
                    }
                }
            } else {
                msgDiv.innerText = text;
            }

            chatContainer.appendChild(msgDiv);
            chatContainer.scrollTop = chatContainer.scrollHeight;
        }

        function handleKeyPress(e) {
            if (e.key === 'Enter') sendMessage();
        }

        async function sendMessage() {
            const text = userInput.value.trim();
            if (!text) return;

            appendMessage(text, 'user');
            userInput.value = '';

            appendMessage("Qwen soch raha hai...", 'bot');
            const loadingMsg = chatContainer.lastChild;

            try {
                const response = await fetch('/api/chat', {
                    method: 'POST',
                    headers: { 'Content-Type': 'application/json' },
                    body: JSON.stringify({ message: text })
                });
                const data = await response.json();

                loadingMsg.remove();
                appendMessage(data.reply, 'bot');
            } catch (error) {
                loadingMsg.innerText = "Error: Connection nahi ban paya.";
            }
        }
    </script>
</body>
</html>
"""

# 3. Flask Backend (Aapka Naya Qwen Model Connecting Route)
@app.route('/')
def home():
    return html_code

@app.route('/api/chat', methods=['POST'])
def chat():
    user_data = request.json
    user_message = user_data.get('message', '')

    try:
        # 👇 Yahan model ka naam badal kar qwen2.5-coder:1.5b kar diya hai
        response = ollama.chat(
            model='qwen2.5-coder:1.5b',
            messages=[{'role': 'user', 'content': user_message}]
        )
        return jsonify({'reply': response['message']['content']})
    except Exception as e:
        return jsonify({'reply': f"Error: Ollama server response nahi de raha. {str(e)}"})

# 4. Web Server Proxy Launch
def run_server():
    app.run(host='0.0.0.0', port=6006)

threading.Thread(target=run_server).start()

from google.colab.output import eval_js
public_url = eval_js("google.colab.kernel.proxyPort(6006)")
print("🔗 Apne Qwen HTML Interface par chat karne ke liye niche diye gaye link par click karein:")
print(public_url)
````

````python

import numpy as np

# 1. हमारा डेटा (X = एरिया, y = असली कीमत)
X = np.array([1000, 1500, 2000, 2500])
y = np.array([50, 75, 100, 125])  # कीमत लाख में

# 2. शुरुआत में स्लोप (m) और इंटरसेप्ट (c) को 0 मान लेते हैं
m = 0.0
c = 0.0

# सीखने की रफ़्तार (Learning Rate) और कितनी बार लूप चलाना है (Epochs)
lr = 0.0000001
epochs = 1000
n = len(X)

# 3. ट्रेनिंग लूप (यही वो कोडिंग है जो AI को सिखाती है)
for i in range(epochs):
    # मॉडल का तुक्का (Prediction)
    y_pred = m * X + c

    # कैलकुलस (Derivatives) का उपयोग करके गलती (Error) निकालना
    # d_m = -2/n * sum(X * (y - y_pred))
    d_m = (-2/n) * sum(X * (y - y_pred))
    d_c = (-2/n) * sum(y - y_pred)

    # स्लोप और इंटरसेप्ट को अपडेट करना
    m = m - lr * d_m
    c = c - lr * d_c

# फाइनल रिजल्ट देखना
print(f"ट्रेनिंग पूरी हुई! कंप्यूटर का ढूंढा गया फॉर्मूला: y = {m:.4f} * X + {c:.4f}")

# नए घर की कीमत का अंदाजा लगाना
new_area = 1800
predicted_price = m * new_area + c
print(f"{new_area} SqFt घर की अनुमानित कीमत होगी: ₹{predicted_price:.2f} लाख")
````

````python

import pickle

# 1. जो गणितीय मान (Weights) कंप्यूटर ने सीखे, उन्हें एक डिक्शनरी में रखना
my_ai_weights = {
    'model_name': 'My Scratch Linear Regression',
    'slope_m': m,
    'intercept_c': c
}

# 2. 'my_scratch_model.pkl' नाम की एक बाइनरी फाइल बनाना और उसमें सेव करना
with open('my_scratch_model.pkl', 'wb') as file:
    pickle.dump(my_ai_weights, file)

print("बधाई हो! आपकी खुद की एआई फ़ाइल 'my_scratch_model.pkl' बन चुकी है।")
````

````python

import numpy as np

# 1. Simulating 4 words embeddings vectors (Size: 4 words x 3 features)
# Words: ["Python", "is", "very", "fast"]
word_embeddings = np.array([
    [0.9, 0.1, 0.2],  # Python
    [0.1, 0.8, 0.1],  # is
    [0.2, 0.7, 0.3],  # very
    [0.8, 0.2, 0.9]   # fast
])

# 2. Simulating the Fusion/Attention scores (Kon sa word kis par kitna dhyan dega)
# Higher value means stronger fusion bond between those pairs
attention_scores = np.array([
    [0.40, 0.10, 0.10, 0.40],  # 'Python' focuses heavily on 'fast'
    [0.10, 0.60, 0.20, 0.10],  # 'is' focuses on grammar
    [0.10, 0.20, 0.60, 0.10],  # 'very' intensifies the next word
    [0.45, 0.05, 0.10, 0.40]   # 'fast' binds deeply back with 'Python'
])

# 3. Fusing the words together using Matrix Multiplication (@ operator)
# This blends the representations of all words based on their context bonds
fused_context = attention_scores @ word_embeddings

print("Original Individual Vectors Matrix:\n", word_embeddings)
print("\nFused Context Matrix (Words combined with their meanings blended):\n", fused_context)
````
